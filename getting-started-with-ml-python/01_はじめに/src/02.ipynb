```python
import sys
import numpy as np
```

```python
py_list = list(range(1000))
np_arr = np.array(range(1000), dtype=np.int64)
```

```python
print(sys.getsizeof(py_list))  # 約 8056 バイト
print(np_arr.nbytes)           # 約 8000 バイト（差は小さいが…）

# 要素1個あたりのコスト
# Pythonリスト: ポインタ8B + オブジェクト28B = 36B/個
# NumPy:        値のみ 8B = 8B/個
```
